```python
import os
from pathlib import Path

print("Current working dir:", Path.cwd())

PROJECT_ROOT = Path.cwd().parent
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

print("Looking in:", DATA_PROCESSED)
print("Files found:", list(DATA_PROCESSED.iterdir()))
```

```python
df = pd.read_csv(DATA_PROCESSED / "HDFCBANK.NS_labeled_2015-01-01_2025-01-01.csv", index_col=0, parse_dates=True)
df.head()
```

```python
df.columns
```

```python
df['sma_10'] = ta.trend.sma_indicator(df['Close'], window=10)
df['sma_20'] = ta.trend.sma_indicator(df['Close'], window=20)
df['sma_50'] = ta.trend.sma_indicator(df['Close'], window=50)
df['ema_10'] = ta.trend.ema_indicator(df['Close'], window=10)
df['ema_20'] = ta.trend.ema_indicator(df['Close'], window=20)
```

```python
df['rsi_14'] = ta.momentum.rsi(df['Close'], window=14)
```

```python
macd = ta.trend.MACD(df['Close'])
df['macd'] = macd.macd()
df['macd_signal'] = macd.macd_signal()
df['macd_diff'] = macd.macd_diff()
```

```python
bb = ta.volatility.BollingerBands(df['Close'], window=20, window_dev=2)
df['bb_high'] = bb.bollinger_hband()
df['bb_low'] = bb.bollinger_lband()
df['bb_width'] = bb.bollinger_wband()
```

```python
df.isna().sum()
```

```python
df = df.dropna()
```

```python
ticker = "HDFCBANK.NS"
start_date = "2015-01-01"
end_date = "2025-01-01"

df.to_csv(DATA_PROCESSED / f"{ticker}_engineered_{start_date}_{end_date}.csv")
```
