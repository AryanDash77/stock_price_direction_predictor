```python
import pandas as pd
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
```

```python
ticker = "HDFCBANK.NS"
start_date = "2015-01-01"
end_date = "2025-01-01"
```

```python
df = pd.read_csv(DATA_PROCESSED / f"{ticker}_engineered_{start_date}_{end_date}.csv", index_col=0, parse_dates=True)
df.head()
```

```python
split_index = int(len(df) * 0.8)

train = df.iloc[:split_index]
test = df.iloc[split_index:]
```

```python
print("Train date range:", train.index.min(), "to", train.index.max())
print("Test date range:", test.index.min(), "to", test.index.max())
print("Train shape:", train.shape, "| Test shape:", test.shape)
```

```python
feature_cols = [
    'Open', 'High', 'Low', 'Close', 'Volume', 'daily_return',
    'sma_10', 'sma_20', 'sma_50', 'ema_10', 'ema_20',
    'rsi_14', 'macd', 'macd_signal', 'macd_diff',
    'bb_high', 'bb_low', 'bb_width'
]
```

```python
X_train = train[feature_cols]
y_train = train['target']

X_test = test[feature_cols]
y_test = test['target']
```
