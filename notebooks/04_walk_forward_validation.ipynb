```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import TimeSeriesSplit
```

```python
PROJECT_ROOT = Path.cwd().parent
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"
```

```python
ticker = "HDFCBANK.NS"
start_date = "2015-01-01"
end_date = "2025-01-01"

df = pd.read_csv(DATA_PROCESSED / f"{ticker}_engineered_{start_date}_{end_date}.csv", index_col=0, parse_dates=True)
```

```python
feature_cols = [
    'Open', 'High', 'Low', 'Close', 'Volume', 'daily_return',
    'sma_10', 'sma_20', 'sma_50', 'ema_10', 'ema_20',
    'rsi_14', 'macd', 'macd_signal', 'macd_diff',
    'bb_high', 'bb_low', 'bb_width'
]
```

```python
X = df[feature_cols]
y = df['target']
```

```python
tscv = TimeSeriesSplit(n_splits = 5)
```

```python
for fold, (train_idx, test_idx) in enumerate(tscv.split(X), start=1):
    train_dates = df.index[train_idx]
    test_dates = df.index[test_idx]
    print(f"Fold {fold}:")
    print(f"  Train: {train_dates.min().date()} to {train_dates.max().date()} ({len(train_idx)} rows)")
    print(f"  Test:  {test_dates.min().date()} to {test_dates.max().date()} ({len(test_idx)} rows)")
```
